## Palabras claves

### Membrana bimodal

**¿Qué significa este comportamiento bimodal?**

Normalmente, las neuronas tienen un solo potencial de membrana en reposo. En cambio, las **neuronas espinosas medias (MSN)** del cuerpo estriado pueden existir en dos estados diferentes:

* **Estado bajo ("down state"):** Un estado hiperpolarizado y estable, donde el potencial de membrana está relativamente bajo y la neurona está inactiva.
* **Estado alto ("up state"):** Un estado despolarizado y estable, donde el potencial de membrana es más alto y la neurona es mucho más propensa a disparar potenciales de acción.

**¿Qué causa este comportamiento?**

La transición entre estos dos estados está influenciada por varios factores, y la modulación dopaminérgica es clave.

* **Modulación dopaminérgica:** La dopamina, a través de sus receptores (especialmente los receptores D1 y D2), influye en la excitabilidad de estas neuronas y en su capacidad para pasar de un estado a otro.
* **Receptores de NMDA:** Un estudio sugiere que la modulación puede ser causada por el bloqueo dependiente del voltaje de los receptores NMDA.
* **Sinapsis:** La entrada de información sináptica excitatoria también juega un papel fundamental en el cambio de estos estados.

**Implicaciones funcionales**

Este comportamiento bimodal es fundamental para la función del cuerpo estriado, una región del cerebro involucrada en el control del movimiento, la toma de decisiones y la formación de hábitos. La capacidad de las neuronas para cambiar entre estos estados permite al cerebro integrar información y seleccionar acciones de manera flexible.

En resumen, la "membrana bimodal" en este contexto no se refiere a una estructura física con dos partes, sino a un **fenómeno dinámico** y **electrofisiológico** en el que las neuronas **pueden mantener dos niveles de actividad diferentes y estables**.

### Cuerpo estriado

**Definición**

El cuerpo estriado también es conocido como ***núcleo estriado*** y ***neoestriado***. Se trata de un conjunto de estructuras localizadas a nivel subcortical, es un grupo de núcleos interconectados que forman la estructura más grande de los ganglios basales, ubicados en la base del cerebro. Su nombre proviene de la apariencia rayada de materia gris y blanca que forman sus estructuras.

**Estructura**

El cuerpo estriado se divide en dos partes principales:

* **Estriado dorsal (o neoestriado):** Compuesto por el núcleo caudado y el putamen. Estos núcleos están separados por la cápsula interna, una banda de fibras nerviosas.
* **Estriado ventral:** Incluye el núcleo accumbens y el tubérculo olfatorio.

imagen_2025-10-11_175432648.png

**Funciones**

El cuerpo estriado es un componente crucial para varias funciones cerebrales complejas, actuando como la principal "puerta de entrada" para el resto de los ganglios basales:

* **Control motor:** Coordina el movimiento voluntario, la postura y la planificación de acciones, así como los movimientos automáticos.
* **Formación de hábitos:** Desempeña un papel esencial en el aprendizaje y la consolidación de comportamientos repetitivos.
* **Motivación y recompensa:** Forma parte del sistema de recompensa del cerebro, procesando estímulos relacionados con la motivación y el placer.
* **Toma de decisiones:** Participa en la toma de decisiones basada en la recompensa y en el valor de las acciones, adenás de las emociones.

### Interneuronas GABAergicas

Las interneuronas GABAérgicas son las principales neuronas inhibitorias del sistema nervioso central, que utilizan el neurotransmisor ácido gamma-aminobutírico (GABA) para regular la actividad de otros circuitos neuronales. A diferencia de las interneuronas colinérgicas, que modulan la actividad, las interneuronas gabaérgicas suelen ser las responsables directas de frenar o detener la activación neuronal.

**Función en diferentes regiones del cerebro**

La función de las interneuronas gabaérgicas varía según su ubicación y los circuitos en los que participan:
* **Corteza cerebral:** Regulan la actividad de las neuronas piramidales para modular procesos cognitivos como el aprendizaje, la atención y la toma de decisiones.
* **Hipocampo:** Son cruciales para la formación de la memoria y la generación de ritmos neuronales.
* **Ganglios basales:** Las neuronas espinosas medianas del estriado y las neuronas del globo pálido son ejemplos de neuronas de proyección gabaérgicas que controlan los circuitos motores.
* **Cerebelo:** Las células de Purkinje, que son gabaérgicas, son las principales neuronas de proyección del cerebelo, inhibiendo los núcleos cerebelosos profundos para coordinar el movimiento.
* **Médula espinal:** Las interneuronas de Renshaw inhiben las motoneuronas, ayudando a regular la fuerza muscular, mientras que otras interneuronas espinales controlan los arcos reflejos.

**Implicación en enfermedades**

Un desequilibrio en la función de las interneuronas gabaérgicas se ha asociado con numerosos trastornos neurológicos y psiquiátricos:
* **Epilepsia:** La disfunción de las interneuronas gabaérgicas puede llevar a una excesiva excitación neuronal, causando convulsiones.
* **Esquizofrenia y autismo:** Se han reportado anomalías en la densidad o función de las interneuronas, especialmente las que expresan parvalbúmina.
* **Ansiedad:** Muchos fármacos ansiolíticos actúan potenciando la función del GABA, lo que reduce la hiperactividad neuronal.
* **Enfermedad de Alzheimer:** La pérdida de la inhibición gabaérgica en etapas tempranas se asocia con la hiperactividad cerebral observada en esta enfermedad.

### Interneuronas colinérgicas

Las interneuronas colinérgicas son un tipo de neuronas que liberan el neurotransmisor acetilcolina (ACh) y desempeñan un papel crucial en la regulación de la actividad de otros circuitos neuronales. Son un componente clave de la fisiología de los ganglios basales y desempeñan un papel fundamental en el aprendizaje, la atención, la motivación y el control motor.

Aunque son escasas (constituyen solo del 1 al 3 % de las neuronas estriatales), estas interneuronas tienen una influencia poderosa sobre las neuronas espinosas medianas (MSN), las principales neuronas de salida del estriado.

**Implicación en enfermedades**

La disfunción de las interneuronas colinérgicas está relacionada con varias patologías neuropsiquiátricas y neurodegenerativas.

* **Enfermedad de Parkinson:** Se cree que la pérdida de dopamina altera el equilibrio entre la dopamina y la acetilcolina en el estriado, lo que provoca una actividad colinérgica anormal y contribuye a los síntomas motores.
* **Adicciones:** La interacción entre los sistemas colinérgicos y dopaminérgicos en el núcleo accumbens se ha implicado en la conducta motivada y la adicción.
* **Enfermedad de Huntington y síndrome de Tourette:** La alteración de estas interneuronas también se ha asociado a la fisiopatología de otras enfermedades neurológicas.

### Neuronas de proyección espinosa media (MSN)

Las neuronas de proyección espinosa media (MSN) son el tipo de neurona predominante en el estriado, una región clave de los ganglios basales en el cerebro. Se les conoce por su tamaño intermedio y por las numerosas espinas en sus dendritas, que actúan como puntos de conexión para otras neuronas.

**Funciones y papel en enfermedades**

* **Control del movimiento:** El equilibrio entre las vías directa e indirecta, mediado por las MSN, es crucial para el control motor normal. Un desequilibrio o disfunción en estas neuronas puede llevar a trastornos del movimiento.
* **Recompensa y motivación:** Las MSN integran señales de recompensa de los circuitos de dopamina, influyendo en la motivación y los comportamientos orientados a metas.
* **Enfermedad de Huntington:** Las MSN son particularmente vulnerables en esta enfermedad neurodegenerativa. La pérdida de estas neuronas en el estriado contribuye a los síntomas motores y cognitivos de la enfermedad.
* **Enfermedad de Parkinson:** La pérdida de la señal de dopamina que modula la actividad de las MSN está implicada en los síntomas motores de la enfermedad de Parkinson.

Los nuevos modelos capturan las relaciones entrada-salida tanto para la inyección actual como para los picos de entrada con notable precisión, a pesar de la disminución de orden de magnitud en el tamaño del sistema. También capturan la facilitación de pulsos emparejados que muestran los MSN.



**Conclusión del paper**

Los modelos de dopamina predicen que los efectos sinápticos dominan los efectos intrínsecos para todos los niveles de activación de los receptores D1 y D2. Se demuestra que esta bimodalidad se crea modelando los efectos agonistas como lentos, saltos irregulares y masivos en la conductancia de NMDA y, más que una forma de biestabilidad, se debe al bloqueo dependiente del voltaje de los receptores de NMDA. Los modelos también predicen una bimodalidad del potencial de membrana más pronunciada después de la activación del receptor D1. Por lo tanto, este trabajo establece modelos reducidos pero precisos de MSN modulados por dopamina, adecuado para su uso en modelos a gran escala del cuerpo estriado. Más importante aún, estos proporcionan un marco manejable para estudios adicionales de los efectos de la dopamina en la computación de neuronas individuales.

## Introducción

La pérdida de células de dopamina en la enfermedad de Parkinson y sus modelos animales conduce a profundos déficits motores. Un sistema de dopamina intacto también parece ser crítico para muchas formas de aprendizaje consistente con las correlaciones reportadas entre la activación de neuronas dopaminérgicas y el error de predicción requerido por las teorías del aprendizaje por refuerzo.

Gran parte del trabajo para comprender estas funciones de la dopamina se ha centrado en el cuerpo estriado, el principal núcleo de entrada de los ganglios basales. El cuerpo estriado es el principal lugar de acción de la dopamina, ya que contiene, con diferencia, la mayor densidad de receptores de dopamina en el cerebro de los vertebrados. Además, el cuerpo estriado recibe aportes convergentes masivos del neocórtex, el tálamo, la formación del hipocampo y la amígdala, y la dopamina modula las respuestas de las neuronas estriatales a ellos. Por tanto, los problemas gemelos de comprender las funciones computacionales de la dopamina y el cuerpo estriado están inseparablemente entrelazados.

Dentro del cuerpo estriado se encuentra un microcircuito complejo, predominantemente GABAérgico. Las neuronas de proyección espinosa media (MSN) son las únicas neuronas de salida y comprenden hasta el 95% de la población de neuronas en el cuerpo estriado de las ratas, mientras que las interneuronas GABAérgicas y colinérgicas forman la mayor parte de la población restante.

La dopamina tiene múltiples efectos diferentes sobre estos tipos de neuronas y a través de múltiples receptores: de hecho, los efectos precisos de la activación de los receptores de dopamina en el MSN han sido muy debatidos. Comprender la contribución de todos estos elementos a la función estriatal idealmente requiere modelos a gran escala que replican los tipos, números y conectividad de neuronas en una escala uno a uno. Construir a tales escalas requiere modelos de neuronas individuales lo suficientemente simples como para ser computacionalmente manejables, pero lo suficientemente complejos como para capturar el comportamiento característico de una especie de neurona.

## Ecuaciones y parámetros del paper

Se usa la forma canónica de Izhikevich para generar potenciales de acción. Con
$v$ el potencial de membrana y $u$ el término de la corriente lenta dominante de la clase neuronal, el modelo es:

**Eqs. (1-2)**

$C\dot{v}=k(v-v_r)(v-v_t)-u+I(t)$ ***[1]***

$\dot{u}=a[b(v−v_r​)−u]$ ***[2]***

**NOTA:** EL PUNTO ENCIMA DE $\dot{v}$ y $\dot{u}$ INDICA LA DERIVADA TEMPORAL:
* $\dot{v}=\frac{dv}{dt}$: Es la velocidad de cambio del voltaje de membrana.
* $\dot{u}=\frac{du}{dt}$: Es la velocidad de cambio de la variable lenta $u$.

Con la condición de reset:

si $v(t)≥v_{peak} ​⇒{v←c,u←u+d}$

* $v←c$: Reinicia el voltaje a un nivel subumbral.
* $u←u+d$: Aumenta la corriente lenta (adaptación), frenando disparos inmediatos y moldeando la f-I.

| Símbolo           | Nombre / Unidad típica                  | Qué controla / Intuición práctica                                                                                                                                                                     |
| ----------------- | --------------------------------------- | ----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| ($v(t)$)            | **Voltaje de membrana** (mV)            | Variable rápida. Lo que graficas como traza.                                                                                                                                                          |
| ($u(t)$)            | **Corriente lenta / recuperación** (pA) | Suma efectiva de procesos lentos (adaptación, canales dependientes de voltaje). Se resta en la ecuación de (v); más (u) ⇒ más freno.                                                                  |
| ($C$)               | **Capacitancia** (pF)                   | Inercia del voltaje: grande ⇒ (v) cambia más **lento**; pequeña ⇒ más **rápido**. Afecta la latencia y la pendiente de f-I.                                                                           |
| ($k$)               | **Factor cuadrático** del I-V (pA/mV²)  | Da forma a la no linealidad de inicio de spike. En este modelo, **reducir (k)** (como en D2) **baja el reobase** y desplaza f-I a la izquierda (más excitable).                                       |
| ($v_r$)             | **Potencial de reposo** (mV)            | Punto de referencia subumbral. Más negativo ⇒ más lejos del umbral ⇒ más corriente y **más latencia** para disparar.                                                                                  |
| ($v_t$)             | **Potencial “umbral”** (mV)             | Junto a (k) determina el reobase. Más negativo ⇒ más fácil disparar.                                                                                                                                  |
| ($I(t)$)            | **Corriente de entrada** (pA)           | Puede ser DC (f-I) o la **suma sináptica** (AMPA+NMDA+GABA) en f-f.                                                                                                                                   |
| ($a$)               | **Velocidad de (u)** ($ms^{-1}$)        | Fija ($\tau_u\approx 1/a$). (a) pequeño ⇒ (u) es muy **lento** (latencias largas; adaptación sostenida).                                                                                                |
| ($b$)               | **Acoplamiento (v $\to$ u)** (pA/mV ≈ nS)  | Define la nulclina lenta ($u_\infty (v)=b(v-v_r)$). Con (b<0) (caso MSN) la meta de (u) **disminuye** al despolarizarse (v); combinado con el salto (d) genera adaptación y forma de f-I característica. |
| ($c$)               | **Voltaje de reinicio** (mV)            | A dónde cae (v) tras un spike. Más bajo ⇒ ISIs más largos (más tiempo para volver a umbral).                                                                                                          |
| ($d$)               | **Incremento de (u) tras spike** (pA)   | “Patada” de adaptación post-spike. Mayor (d) ⇒ más freno tras cada disparo ⇒ **menor** frecuencia sostenida (reduce pendiente de f-I).                                                                |
| ($v_{\text{peak}}$) | **Umbral de corte** (mV)                | Valor a partir del cual declaras un spike y aplicas el reset; no afecta la dinámica subumbral.                                                                                                        |


De las ecuaciones ***[1]*** y ***[2]*** se deducen las nulclinas:
* Nulclina de v: $u=k(v-v_r)(v-v_t)+I(t)$
* Nulclina de u: $u=b(v-v_r)$

Para el marco de dopamina (niveles de activación) se instroducen factores de activacion $\phi_1$ (Receptores D1) y $\phi_2$ (Receptores D2). Durante el ajuste se aume que $\phi_1=\phi_2=0.8$ para aproximar activación casi máxima.

Para la modulación instrínseca por dopamina se tienen dos vías:

* ***Vía D1 (hiperpolarizante a nivel intrínseco)***:
se implementa como cambio en dos parámetros del modelo básico:
  * Desplazamiento del potencial de reposo: $v_r ← v_r(1+k \phi_1)$ ***[3]***. Si $k>0$ escala cuánto se desplaza la posición de reposo $v_r$ con D1.
  * Reducción del incremento de la corriente lenta (recuperación) tras el spike: $d ← d(1-L\phi_1)$ ***[4]***. Si $L>0$ escala cuánto redulce el salto de adaptación $d$ tras cada spike bajo D1.

**NOTA: Estas sustituciones se aplican en las ecuaciones del modelo.**

* ***Vía D2 (facilitadora a nivel intrínseco)***:
  * Reducción de la ganancia $k$: $k ← k(1-α \phi_2)$ ***[5]***. Si $α>0$ escala la reducción de la ganancia cuadrática $k$ del término $k(v-v_r)(v-v_t)$.

Para el modelo de la corriente sináptica de entrada se usa la siguiente ecuación.

$I_{syn}=I_{ampa}+B(v)I_{nmda}+I_{gaba}$ ***[6]***

Cada entrada sináptica sub z (donde $z$ es uno de los siguientes: $ampa$, $nmda$, $gaba$) y es modelada por:

$I_z=\overline{g}h_z(E_z-v)$ ***[7]***

Donde:
* $\overline{g}$ es la conductancia máxima.
* $E_z$ es el potencial de reversión.
* $h_z$ es la variable de apertura sináptica.

$\dot{h}_z=\frac{-h_z}{τ_z}$ y $h_z(t) ← \frac{h_z(t)+S_z(t)}{τ_z}$ ***[8]***

Donde:
* $τ_z$ es la constante sináptica de tiempo.
* $S_z(t)$ es el numero de spikes que cayeron en ese paso de tiempo $t$.

At each time-step $Δt$ (seconds), and given a mean spike rate $r$ (spikes/s), we compute the probability of a spike per afferent as $p(s) = rΔt$. The total number of spike-events $S$ at each time-step is then just drawn from a binomial distribution $S = B(N, p(s))$.

El término $B(v)$ modela el bloque de NMDA por $Mg^{2+}$, modela el enchufe de magnesio dependiente del voltaje en los receptores NMDA descrito por la siguiente ecuación:

$B(v)=\frac{1}{1+ \frac{[Mg^{2+}]_0}{3.57}exp(-0.062v)}$ ***[9]***

Donde $[Mg^{2+}]_0$ es la concentración de equilibrio de iones de magnesio.

Para la modulación de dopamina en corrientes sinápticas de **NMDA** se modela con la siguiente ecuación:

$I_{NMDA}^{D1}=I_{NMDA}(1+ β_1 \phi_1)$ ***[10]***

Donde $β_1$ es la magnitud del efecto D1 sobre **NMDA**.

Para la modulación de dopamina en corrientes sinápticas de AMPA se modela con la siguiente ecuación:

$I_{AMPA}^{D2}=I_{AMPA}(1- β_2 \phi_2)$ ***[11]***

Donde $β_2$ es la magnitud del efecto D1 sobre **AMPA**.

## DEFINICION DE FUNCIONES

```python
def dv_dt(C, k, v, v_r, v_t, u, I_syn): # Ecuacion 1
  return (k*(v - v_r)*(v - v_t) - u + I_syn)/C


def du_dt(a,b, v, v_r, u): # Ecuacion 2
  return a*(b*(v - v_r) - u)


# Ecuaciones 7
def I_ampa(v, h_ampa):
  return g_ampa*h_ampa*(E_ampa - v)

def I_nmda(v, h_nmda):
  return g_nmda*h_nmda*(E_nmda - v)

def I_gaba(v, h_gaba):
  return g_gaba*h_gaba*(E_gaba - v)


''' La idea es modelar como cambia h en funcion del cambio en el tiempo, es por eso que en realidad se modela:
                              h_z(t + delta(t)) = h_z(t)*e^{-delta(t)/tau_z}
    Lo que significa que:
                              h_{z+1} = h_z*e^{-delta(t)/tau_z} + S_n/tau_z
    De ahí el por qué aparece el np.exp
'''
def actualizar_h(h, tau, S, dt): # Ecuacion 8
  return h*np.exp(-dt/tau) + (S / tau)


def B(v, Mg_0): # Ecuacion 9
  return 1/(1 + (Mg_0/3.57)*np.exp(-0.062*v))
```

## SOLUCION POR EULER GRAFICA 2A.

```python
corrientes = np.arange(220, 300.1, 5)

''' El paper aclara que los resultados de las graficas los obtiene en los ultimos 4 segundos, toca aumentar la duración final
    El paper mide con inyección de 5 s y calcula la tasa en la ventana final de 4 s. Las MSNs tienen latencias largas cerca del umbral (cientos de ms)
'''
tf = 5000 # ms
ventana_final = 4000 # ms (Ultimos 4 seg)

N = int(tf/pasos) # Numero de pasos

# Vector con los tiempos
tiempoBase = np.arange(0, tf, pasos)
tiempoD1 = np.arange(0, tf, pasos)
tiempoD2 = np.arange(0, tf, pasos)

# Vector con las frecuencias de disparo para cada corriente
frecuenciasBase = []
frecuenciasD1 = []
frecuenciasD2 = []

h_ampaBase = h_ampa
h_nmdaBase = h_nmda
h_gabaBase = h_gaba
h_ampaD1 = h_ampa
h_nmdaD1 = h_nmda
h_gabaD1 = h_gaba
h_ampaD2 = h_ampa
h_nmdaD2 = h_nmda
h_gabaD2 = h_gaba
```

### Baseline

```python
''' Solucion para base-line '''
for corriente in corrientes:
  vBase = [v0]
  uBase = [u0]

  #Vector con los tiempos donde la nuerona dispara
  tiempos_spikesBase = []

  for i in range(N-1):
    vOldBase = vBase[-1]
    uOldBase = uBase[-1]

    I_synBase = corriente

    vNewBase = vOldBase + dv_dt(C, k, vOldBase, v_r, v_t, uOldBase, I_synBase) * pasos
    uNewBase = uOldBase + du_dt(a, b, vOldBase, v_r, uOldBase) * pasos

    if vNewBase >= v_peak:
      vNewBase = c
      uNewBase += d
      tiempos_spikesBase.append(tiempoBase[i])

    vBase.append(vNewBase)
    uBase.append(uNewBase)

  t0 = tf - ventana_final
  sp = [t for t in tiempos_spikesBase if t >= t0]
  frecuenciasBase.append(len(sp) / (ventana_final/1000))  # Calcula que tan a menudo ocurren los spikes en relación con el tiempo transcurrido.
```

### Dopamine Intrinsic 1 (D1)

```python
''' Solución para Dopamine Intrinsinc 1 (D1) '''
for corriente in corrientes:
  vD1 = [v0]
  uD1 = [u0]

  #Vector con los tiempos donde la nuerona dispara
  tiempos_spikesD1 = []

  for i in range(N-1):
    vOldD1 = vD1[-1]
    uOldD1 = uD1[-1]

    v_rD1 = v_r*(1 + K*phi_1) # Ecuacion 3
    d_D1 = d*(1 - L*phi_1) # Ecuacion 4

    I_synD1 = corriente

    vNewD1 = vOldD1 + dv_dt(C, k, vOldD1, v_rD1, v_t, uOldD1, I_synD1) * pasos
    uNewD1 = uOldD1 + du_dt(a, b, vOldD1, v_rD1, uOldD1) * pasos

    if vNewD1 >= v_peak:
      vNewD1 = c
      uNewD1 += d_D1
      tiempos_spikesD1.append(tiempoD1[i])

    vD1.append(vNewD1)
    uD1.append(uNewD1)

  t0 = tf - ventana_final
  sp = [t for t in tiempos_spikesD1 if t >= t0]
  frecuenciasD1.append(len(sp) / (ventana_final/1000.0))
```

### Dopamine Intrinsic 2 (D2)

```python
''' Solución para Dopamine Intrinsinc 2 (D2) '''
for corriente in corrientes:
  vD2 = [v0]
  uD2 = [u0]

  #Vector con los tiempos donde la nuerona dispara
  tiempos_spikesD2 = []

  for i in range(N-1):
    vOldD2 = vD2[-1]
    uOldD2 = uD2[-1]

    kD2 = k*(1 - alpha*phi_2) # Ecuacion 5

    I_synD2 = corriente

    vNewD2 = vOldD2 + dv_dt(C, kD2, vOldD2, v_r, v_t, uOldD2, I_synD2) * pasos
    uNewD2 = uOldD2 + du_dt(a, b, vOldD2, v_r, uOldD2) * pasos

    if vNewD2 >= v_peak:
      vNewD2 = c
      uNewD2 += d
      tiempos_spikesD2.append(tiempoD2[i])

    vD2.append(vNewD2)
    uD2.append(uNewD2)

  t0 = tf - ventana_final
  sp = [t for t in tiempos_spikesD2 if t >= t0]
  frecuenciasD2.append(len(sp) / (ventana_final/1000.0))
```

## SOLUCION POR EULER CON CORRIENTE SINAPTICA GRAFICA 2C.

```python
valores_r = np.linspace(4.0, 8.0, 11)   #  Finally, we varied the mean rate r of each of the N underlying spike trains between 4 and 8 spikes/s. n = 11 puntos deducidos del paper
total_inputs = 168.0 * valores_r        # S* = 2*N*r con N=84 por tipo (168 synapses, 84 GABA and 84 glutamate)

tf = 5000 # ms
ventana_final = 4000 # ms (Ultimos 4 seg)

N = int(tf/pasos) # Numero de pasos

# Vector con los tiempos
tiempoBase = np.arange(0, tf, pasos)
tiempoD1 = np.arange(0, tf, pasos)
tiempoD2 = np.arange(0, tf, pasos)
tiempoD1comp = np.arange(0, tf, pasos)
tiempoD2comp = np.arange(0, tf, pasos)

# Vector con las frecuencias de disparo para cada corriente
frecuenciasBase = []
frecuenciasD1 = []
frecuenciasD2 = []
frecuenciasD1comp = []
frecuenciasD2comp = []

h_ampaBase = h_ampa
h_nmdaBase = h_nmda
h_gabaBase = h_gaba
h_ampaD1 = h_ampa
h_nmdaD1 = h_nmda
h_gabaD1 = h_gaba
h_ampaD2 = h_ampa
h_nmdaD2 = h_nmda
h_gabaD2 = h_gaba
h_ampaD1comp = h_ampa
h_nmdaD1comp = h_nmda
h_gabaD1comp = h_gaba
h_ampaD2comp = h_ampa
h_nmdaD2comp = h_nmda
h_gabaD2comp = h_gaba

rng = np.random.default_rng(1)  # Le asigno una semilla para tener siempre el mismo resultado

''' S modela el numero de spikes que cayeron en un paso de tiempo t
    En este modelo, los spikes presinápticos se asumen procesos aleatorios (Poisson) que llegan desde fuera;
    no estamos simulando el voltaje de 168 neuronas presinápticas para luego “detectarles picos”, por eso no se usa 'find_peaks'.
    Por tal motivo se generan por muestreo aleatorio con rng, pero no es “al azar”, es exactamente la distribución que el modelo prescribe, una distribución de Poisson.

    Se puede usar una binomial que garantiza S ≤ N (a lo sumo 1 spike por aferente y bin). Poisson garantiza cuenta total aunque
    puede producir >N si dt no es muy pequeño.
'''
# Para Poisson o binomial se debe aplicar con segundos, y como los pasos están en ms, por eso la conversion
dt_s = pasos/1000
```

#### Primer spike sináptico vs frecuencia de entrada (3F)

```python
phi = 0.8
tf3F = 3000
N = int(tf3F/pasos) # Numero de pasos

def latencia_d2(r, num_ensayos=100):
    latencias = []

    for m in range(num_ensayos):
        h_ampaD2comp = h_ampa
        h_nmdaD2comp = h_nmda
        h_gabaD2comp = h_gaba

        vD2comp = [v0]
        uD2comp = [u0]
        tiempo_primer_spike = None

        for i in range(N-1):
            kD2comp = k * (1 - alpha * phi)  # Ecuacion 5

            vOldD2comp = vD2comp[-1]
            uOldD2comp = uD2comp[-1]

            S_plusD2comp  = rng.binomial(84, r * dt_s)
            S_minusD2comp = rng.binomial(84, r * dt_s)

            h_ampaD2comp = actualizar_h(h_ampaD2comp, tau_ampa, S_plusD2comp, pasos)
            h_nmdaD2comp = actualizar_h(h_nmdaD2comp, tau_nmda, S_plusD2comp, pasos)
            h_gabaD2comp = actualizar_h(h_gabaD2comp, tau_gaba, S_minusD2comp, pasos)

            ID2_ampa = I_ampa(vOldD2comp, h_ampaD2comp) * (1 - beta_2 * phi)  # Ecuacion 11
            I_synD2comp = (ID2_ampa +
                          B(vOldD2comp, Mg_0) * I_nmda(vOldD2comp, h_nmdaD2comp) +
                          I_gaba(vOldD2comp, h_gabaD2comp))

            vNewD2comp = vOldD2comp + dv_dt(C, kD2comp, vOldD2comp, v_r, v_t, uOldD2comp, I_synD2comp) * pasos
            uNewD2comp = uOldD2comp + du_dt(a, b, vOldD2comp, v_r, uOldD2comp) * pasos

            if vNewD2comp >= v_peak:
                tiempo_primer_spike = i * pasos
                break

            vD2comp.append(vNewD2comp)
            uD2comp.append(uNewD2comp)

        if tiempo_primer_spike:
            latencias.append(tiempo_primer_spike)
    return np.mean(latencias)

def latencia_d1(r, num_ensayos=100):
    latencias = []

    for _ in range(num_ensayos):
        h_ampaD1comp = h_ampa
        h_nmdaD1comp = h_nmda
        h_gabaD1comp = h_gaba

        vD1comp = [v0]
        uD1comp = [u0]
        tiempo_primer_spike = None

        for i in range(N-1):
            v_r_D1 = v_r * (1 + K * phi)  # Ecuación 3 - hiperpolarización
            d_D1 = d * (1 - L * phi)      # Ecuación 4 - facilitación

            vOldD1comp = vD1comp[-1]
            uOldD1comp = uD1comp[-1]

            S_plusD1comp = rng.binomial(84, r * dt_s)
            S_minusD1comp = rng.binomial(84, r * dt_s)

            h_ampaD1comp = actualizar_h(h_ampaD1comp, tau_ampa, S_plusD1comp, pasos)
            h_nmdaD1comp = actualizar_h(h_nmdaD1comp, tau_nmda, S_plusD1comp, pasos)
            h_gabaD1comp = actualizar_h(h_gabaD1comp, tau_gaba, S_minusD1comp, pasos)

            I_nmda_D1 = I_nmda(vOldD1comp, h_nmdaD1comp) * (1 + beta_1 * phi)  # Ecuación 10
            I_synD1comp = (I_ampa(vOldD1comp, h_ampaD1comp) +
                          B(vOldD1comp, Mg_0) * I_nmda_D1 +
                          I_gaba(vOldD1comp, h_gabaD1comp))

            vNewD1comp = vOldD1comp + dv_dt(C, k, vOldD1comp, v_r_D1, v_t, uOldD1comp, I_synD1comp) * pasos
            uNewD1comp = uOldD1comp + du_dt(a, b, vOldD1comp, v_r_D1, uOldD1comp) * pasos

            if vNewD1comp >= v_peak:
                vNewD1comp = c
                uNewD1comp += d_D1  # Usar d modificado para D1
                tiempo_primer_spike = i * pasos
                break

            vD1comp.append(vNewD1comp)
            uD1comp.append(uNewD1comp)

        if tiempo_primer_spike:
            latencias.append(tiempo_primer_spike)

    return np.mean(latencias)

def latencia_base(r, num_ensayos=100):
    latencias = []

    for _ in range(num_ensayos):
        h_ampabase = h_ampa
        h_nmdabase = h_nmda
        h_gababase = h_gaba

        vbase = [v0]
        ubase = [u0]
        tiempo_primer_spike = None

        for i in range(N-1):
            vOldbase = vbase[-1]
            uOldbase = ubase[-1]

            S_plusbase = rng.binomial(84, r * dt_s)
            S_minusbase = rng.binomial(84, r * dt_s)

            h_ampabase = actualizar_h(h_ampabase, tau_ampa, S_plusbase, pasos)
            h_nmdabase = actualizar_h(h_nmdabase, tau_nmda, S_plusbase, pasos)
            h_gababase = actualizar_h(h_gababase, tau_gaba, S_minusbase, pasos)

            I_ampabase = I_ampa(vOldbase, h_ampabase)
            I_synbase = (I_ampabase +
                        B(vOldbase, Mg_0) * I_nmda(vOldbase, h_nmdabase) +
                        I_gaba(vOldbase, h_gababase))

            vNewbase = vOldbase + dv_dt(C, k, vOldbase, v_r, v_t, uOldbase, I_synbase) * pasos
            uNewbase = uOldbase + du_dt(a, b, vOldbase, v_r, uOldbase) * pasos

            if vNewbase >= v_peak:
                tiempo_primer_spike = i * pasos
                break

            vbase.append(vNewbase)
            ubase.append(uNewbase)

        if tiempo_primer_spike:
            latencias.append(tiempo_primer_spike)

    return np.mean(latencias)

frecuencias = np.arange(800, 1401, 100)
resultado_D2 = [latencia_d2(f/168, num_ensayos=500) for f in frecuencias]
resultado_D1 = [latencia_d1(f/168, num_ensayos=500) for f in frecuencias]
resultado_base = [latencia_base(f/168, num_ensayos=500) for f in frecuencias]
```

```python
plt.figure(figsize=(8,5))
plt.plot(frecuencias, resultado_D2, 'bs-', linewidth=2, markersize=6, label='D2 (φ=0.8)')
plt.plot(frecuencias, resultado_D1, 'gs-', linewidth=2, markersize=6, label='D1 (φ=0.8)')
plt.plot(frecuencias, resultado_base, 'rs-', linewidth=2, markersize=6, label='base (sin dopamina)')
plt.xlabel('input rate (frecuencia de eventos de entrada)')
plt.ylabel('latencia al primer spike (ms)')
plt.title('Primer spike sináptico vs input rate (frecuencia)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

### Facilitación del pulso pareado (4A)

```python
a = 0.01

t_on_1 = 50.0   # Inicio del primer pulso
t_off_1 = 250.0 # Fin del primer pulso

t_on_2 = 450.0  # Inicio del segundo pulso
t_off_2 = 650.0 # Fin del segundo pulso

I_amp = 400     # Amplitud del pulso

tf = 1000

# Variables para guardar el tiempo que tarda en generarse un spike, luego de inyectar corriente
t1 = 0
t2 = 0

pasos = 0.1
N = int(tf/pasos)
tiempoBase = np.arange(0, tf, pasos)

vBase = [v0]
uBase = [u0]

for i in range(N-1):
  t = tiempoBase[i]
  vOldBase = vBase[-1]
  uOldBase = uBase[-1]

  if (t_on_1 <= t < t_off_1) or (t_on_2 <= t < t_off_2):
    I_synBase = I_amp
  else:
    I_synBase = 0

  vNewBase = vOldBase + dv_dt(C, k, vOldBase, v_r, v_t, uOldBase, I_synBase) * pasos
  uNewBase = uOldBase + du_dt(a, b, vOldBase, v_r, uOldBase) * pasos

  if vNewBase >= v_peak:
    vNewBase = c
    uNewBase += d

    # Determinar el tiempo que tarda en generase cada spike luego de inyectar corriente
    if (t_on_1 <= t < t_off_1) and (t1 == 0):
      t1 = t
    if (t_on_2 <= t < t_off_2) and (t2 == 0):
      t2 = t

  vBase.append(vNewBase)
  uBase.append(uNewBase)

t1 = abs(t_on_1 - t1)
t2 = abs(t_on_2 - t2)

print(f'Tiempo que tardó en generarse el primer spike: t1 = {t1}')
print(f'Tiempo que tardó en generarse el segundo spike: t2 = {t2}')
print(f't = {t1 - t2}')

plt.figure(figsize=(5,5))
plt.axvline(t_on_1, linestyle='--')
plt.axvline(t_off_1, linestyle='--')
plt.axvline(t_on_2, linestyle='--')
plt.axvline(t_off_2, linestyle='--')
plt.plot(tiempoBase, vBase, label='Base', c='black')
plt.ylim(-85, 20)
plt.legend()

corriente = np.zeros_like(tiempoBase)
for i, t in enumerate(tiempoBase):
    if (t_on_1 <= t < t_off_1) or (t_on_2 <= t < t_off_2):
        corriente[i] = I_amp
    else:
        corriente[i] = 0
plt.figure(figsize=(6,2))
plt.plot(tiempoBase, corriente)
plt.title('Corriente inyectada')
plt.xlabel('Tiempo (ms)')
plt.ylabel('I (pA)')
plt.ylim(-10, I_amp + 50)
plt.grid(True)
plt.show()
```

### Intérvalo entre pulsos (4B)

```python
def pared_pulsed(a=0.01, inicio_pulso_2=400):
  t_on_1 = 0
  t_off_1 = 200

  t_on_2 = inicio_pulso_2
  t_off_2 = inicio_pulso_2 + 200

  I_amp = 400     # Amplitud del pulso

  tf = 2000

  # Variables para guardar el tiempo que tarda en generarse un spike, luego de inyectar corriente
  t1 = 0
  t2 = 0

  pasos = 0.1
  N = int(tf/pasos)
  tiempoBase = np.arange(0, tf, pasos)

  vBase = [v0]
  uBase = [u0]

  for i in range(N-1):
    t = tiempoBase[i]
    vOldBase = vBase[-1]
    uOldBase = uBase[-1]

    if (t_on_1 <= t < t_off_1) or (t_on_2 <= t < t_off_2):
      I_synBase = I_amp
    else:
      I_synBase = 0

    vNewBase = vOldBase + dv_dt(C, k, vOldBase, v_r, v_t, uOldBase, I_synBase) * pasos
    uNewBase = uOldBase + du_dt(a, b, vOldBase, v_r, uOldBase) * pasos

    if vNewBase >= v_peak:
      vNewBase = c
      uNewBase += d

      # Determinar el tiempo que tarda en generase cada spike luego de inyectar corriente
      # El and con el t1 == 0 y t2 == 0, es porque luego del primer spike t1 y t2 ya no son 0, así sólo queda guardado el tiempo del primer spike
      if (t_on_1 <= t < t_off_1) and (t1 == 0):
        t1 = t
      if (t_on_2 <= t < t_off_2) and (t2 == 0):
        t2 = t

    vBase.append(vNewBase)
    uBase.append(uNewBase)

  t1 = abs(t_on_1 - t1)
  t2 = abs(t_on_2 - t2)

  return t1 - t2

intervalos = np.arange(200, 1000.1, 100)

delta_t_normal = []  # Para a = 0.01
delta_t_lento = []   # Para a = 0.02

for interval in intervalos:
  # +200 porque el pulso dura 200ms, entonces el segundo pulso empieza en el intérvalo +200ms.
  diferencia_norm = pared_pulsed(a=0.01,
    inicio_pulso_2=interval+200)
  diferencia_lenta = pared_pulsed(a=0.02,
    inicio_pulso_2=interval+200)
  delta_t_normal.append(diferencia_norm)
  delta_t_lento.append(diferencia_lenta)

plt.figure(figsize=(8, 6))
plt.plot(intervalos, delta_t_normal, 'ro-', linewidth=2, markersize=8, label='a = 0.01 (normal)')
plt.plot(intervalos, delta_t_lento, 'ko-', linewidth=2, markersize=8, label='a = 0.02 (lento)')
plt.xlabel('Inter-pulse interval (ms)', fontsize=12)
plt.ylabel('Δt = t₁ - t₂ (ms)', fontsize=12)
plt.title('Figura 4B: Facilitación de Pulso Pareado', fontsize=14)
plt.legend(fontsize=11)
plt.xlim(0,1200)
plt.ylim(-0.5,40)
plt.show()
```
